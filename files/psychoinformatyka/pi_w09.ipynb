```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
teksty = [
    "Jestem spokojny dzisiaj", "Nie jestem spokojny dzisiaj",
    "Czuję napięcie przed egzaminem", "Po egzaminie odczuwam ulgę",
    "Martwię się wynikami", "Dzisiaj spałem dobrze"
]
print("Liczba tekstów:", len(teksty))
```

```python
wektoryzator = TfidfVectorizer(token_pattern=r"(?u)\b\w+\b")
X = wektoryzator.fit_transform(teksty)
print("Macierz TF-IDF:", X.shape)
```

```python
podobienstwo = cosine_similarity(X)
print("Podobieństwo zdań 1 i 2:", round(podobienstwo[0, 1], 3))
```

```python
X2 = TfidfVectorizer(ngram_range=(1, 2)).fit_transform(teksty)
podobienstwo2 = cosine_similarity(X2)
print("Po dodaniu bigramów:", round(podobienstwo2[0, 1], 3))
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(6, 4))
im = ax.imshow(podobienstwo2, vmin=0, vmax=1)
ax.set(xlabel="Wypowiedź", ylabel="Wypowiedź", title="TF-IDF: unigramy i bigramy")
fig.colorbar(im, ax=ax, label="Podobieństwo cosinusowe")
plt.tight_layout()
plt.show()
```
